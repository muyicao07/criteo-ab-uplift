# Exposure as an endogenous treatment

The Wald ratio uses assignment as an instrument. The feature imbalance calls the random-assignment assumption into question, so the ratio is descriptive of the released sample. A CACE/ATT interpretation would require the assumptions listed below.

Results are loaded from the saved analysis outputs.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Image, Markdown
ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
RESULTS = ROOT / 'results/full-run'
def table(name):
    frame = pd.read_csv(RESULTS / name)
    for column in frame.columns:
        if column == "two_sided_p_value":
            frame[column] = frame[column].map(lambda x: "< 1e-300" if x < 1e-300 else format(x, ".3g"))
        elif column.startswith("expected_loss_"):
            frame[column] = frame[column].map(lambda x: format(x, ".3e"))
    display(frame)
def figure(name):
    display(Image(filename=str(RESULTS / "figures" / name)))
def record(name):
    return json.loads((RESULTS / name).read_text())
```

```python
iv = record("ab_results.json")["iv_exploratory"]
display(Markdown(iv["interpretation"]))
display(iv["required_assumptions"])
table("iv.csv")
```
